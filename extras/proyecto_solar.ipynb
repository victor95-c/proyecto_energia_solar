### Singleton
“Singleton significa que solo quiero tener un objeto de esta clase. En mi proyecto lo uso para la configuración de la API. Así, todos los módulos utilizan la misma configuración y no se crean varias configuraciones diferentes.”

**Justificación vs Alternativas:** ¿Por qué no usar variables globales o módulos estáticos? El patrón Singleton nos da mayor control sobre la inicialización (carga perezosa), permite manejar estado mutable de forma segura si es necesario en un entorno concurrente, y facilita realizar pruebas unitarias inyectando un mock del Singleton en lugar de depender de variables globales sueltas.

[![](https://img.plantuml.biz/plantuml/svg/HOv12W8n34NtXTwXon7n3YEkt7c34Deu1DfqI5jSoDpTLYncb_TV-5bBHQqjHMkiyH5BSTV7_PRboPjxM-FwN1omT4yyuoIuCpIDSst20GEGa6v_wKShfPqKQrCSBICxEm2X5y3YOpbDbdAfYWh-MzzrprcM7X_ZRYK9_T4F)](https://editor.plantuml.com/uml/HOv12W8n34NtXTwXon7n3YEkt7c34Deu1DfqI5jSoDpTLYncb_TV-5bBHQqjHMkiyH5BSTV7_PRboPjxM-FwN1omT4yyuoIuCpIDSst20GEGa6v_wKShfPqKQrCSBICxEm2X5y3YOpbDbdAfYWh-MzzrprcM7X_ZRYK9_T4F)

### correcto 

```python
class APIConfig:
    """
    Patrón Singleton.
    Solo puede existir una instancia de esta clase.
    """

    _instancia = None

    def __new__(cls):
        if cls._instancia is None:
            cls._instancia = super().__new__(cls)

            cls._instancia.api_url = (
                "https://api.open-meteo.com/v1/forecast"
            )

            cls._instancia.unidad_energia = "Wh"
            cls._instancia.unidad_temperatura = "°C"

        return cls._instancia

    def mostrar_configuracion(self):
        return {
            "API": self.api_url,
            "Energía": self.unidad_energia,
            "Temperatura": self.unidad_temperatura
        }
```

## Prototype
"El patrón Prototype permite crear nuevos objetos a partir de un objeto existente. En este caso, tenemos un sistema solar original y mediante el método clonar() hacemos una copia utilizando deepcopy. De esta manera podemos crear otro sistema solar con las mismas características iniciales y después modificarlo sin afectar al original."

**Justificación vs Alternativas:** ¿Por qué no simplemente instanciar con `SistemaSolar(...)` cada vez? Si la inicialización del objeto base fuera costosa (ej. requiere consultas a base de datos o cálculos pesados iniciales), instanciar desde cero sería ineficiente. Prototype copia el estado actual en memoria instantáneamente.

```python
class SistemaSolar:

    def __init__(
        self,
        nombre,
        cantidad_paneles,
        potencia_panel,
        tipo_panel,
        ubicacion
    ):
        self.nombre = nombre
        self.cantidad_paneles = cantidad_paneles
        self.potencia_panel = potencia_panel
        self.tipo_panel = tipo_panel
        self.ubicacion = ubicacion


sistema_original = SistemaSolar(
    "Sistema Solar Ciénaga",
    10,
    550,
    "Monocristalino",
    "Ciénaga, Magdalena"
)

sistema_copia = SistemaSolar(
    "Sistema Solar Santa Marta",
    10,
    550,
    "Monocristalino",
    "Santa Marta, Magdalena"
)
```

### correcto

```python
import copy


class SistemaSolar:
    """
    Patrón Prototype.
    Permite clonar un sistema solar existente.
    """

    def __init__(
        self,
        nombre,
        cantidad_paneles,
        potencia_panel,
        tipo_panel,
        ubicacion
    ):
        self.nombre = nombre
        self.cantidad_paneles = cantidad_paneles
        self.potencia_panel = potencia_panel
        self.tipo_panel = tipo_panel
        self.ubicacion = ubicacion

    def clonar(self):
        return copy.deepcopy(self)

    def calcular_potencia(self):
        return (
            self.cantidad_paneles
            * self.potencia_panel
        )

    def mostrar_informacion(self):
        return (
            f"Nombre: {self.nombre}\n"
            f"Paneles: {self.cantidad_paneles}\n"
            f"Potencia por panel: {self.potencia_panel} W\n"
            f"Tipo: {self.tipo_panel}\n"
            f"Ubicación: {self.ubicacion}\n"
            f"Potencia total: "
            f"{self.calcular_potencia()} W"
        )
```

## Composite 
"El patrón Composite me permite tratar un panel solar individual y un conjunto de paneles mediante una misma interfaz. ComponenteSolar es el componente común, PanelSolar es la hoja y ArregloSolar es el Composite. El arreglo puede contener varios paneles e incluso otros arreglos, y todos pueden calcular su potencia mediante el mismo método calcular_potencia()."


**Justificación vs Alternativas:** ¿Por qué no tener dos listas separadas, una para `paneles_individuales` y otra para `arreglos_de_paneles`? Eso nos obligaría a escribir bucles anidados e `if`s constantes para recorrer la jerarquía, violando el principio Open/Closed. Composite simplifica el cliente (la API) al interactuar con paneles y arreglos uniformemente.

### Incorrecto

```python
class PanelSolar:

    def __init__(self, identificador, potencia, tipo):
        self.identificador = identificador
        self.potencia = potencia
        self.tipo = tipo


class ArregloSolar:

    def __init__(self, nombre):
        self.nombre = nombre
        self.paneles = []

    def agregar(self, panel):
        self.paneles.append(panel)

    def calcular_potencia(self):
        total = 0

        for panel in self.paneles:
            total += panel.potencia

        return total

    def mostrar(self):
        print(f"Arreglo: {self.nombre}")

        for panel in self.paneles:
            print(
                f"Panel {panel.identificador}: "
                f"{panel.potencia} W - {panel.tipo}"
            )

        print(f"Potencia total: {self.calcular_potencia()} W")
```

### Correcto

```python
from abc import ABC, abstractmethod


class ComponenteSolar(ABC):

    @abstractmethod
    def calcular_potencia(self):
        pass

    @abstractmethod
    def mostrar(self):
        pass


class PanelSolar(ComponenteSolar):
    """
    Hoja del patrón Composite.
    """

    def __init__(self, identificador, potencia, tipo):
        self.identificador = identificador
        self.potencia = potencia
        self.tipo = tipo

    def calcular_potencia(self):
        return self.potencia

    def mostrar(self):
        return (
            f"Panel {self.identificador}: "
            f"{self.potencia} W - {self.tipo}"
        )


class ArregloSolar(ComponenteSolar):
    """
    Composite.
    Puede contener varios paneles.
    """

    def __init__(self, nombre):
        self.nombre = nombre
        self.componentes = []

    def agregar(self, componente):
        self.componentes.append(componente)

    def eliminar(self, componente):
        if componente in self.componentes:
            self.componentes.remove(componente)

    def calcular_potencia(self):
        total = 0

        for componente in self.componentes:
            total += componente.calcular_potencia()

        return total

    def mostrar(self):
        resultado = []

        resultado.append(f"\nArreglo: {self.nombre}")

        for componente in self.componentes:
            resultado.append(componente.mostrar())

        resultado.append(
            f"Potencia total: {self.calcular_potencia()} W"
        )

        return "\n".join(resultado)
```

## Decorator 
“Utilicé Decorator para agregar factores que afectan la generación solar sin modificar la clase GeneracionBase. Así puedo agregar pérdidas, eficiencia o temperatura de manera independiente y combinarlos según sea necesario.”

**Justificación vs Alternativas:** ¿Por qué no usar herencia tradicional (ej. `GeneracionConPerdidas`, `GeneracionConPerdidasYEficiencia`)? Eso provocaría una explosión combinatoria de clases. Decorator permite combinar múltiples factores dinámicamente en tiempo de ejecución sin crear cientos de subclases.

### Incorrecto

```python
class GeneracionBase:

    def __init__(
        self,
        potencia,
        horas_sol,
        perdidas=0,
        eficiencia=100,
        temperatura=25
    ):
        self.potencia = potencia
        self.horas_sol = horas_sol
        self.perdidas = perdidas
        self.eficiencia = eficiencia
        self.temperatura = temperatura

    def calcular(self):

        energia = self.potencia * self.horas_sol

        # Pérdidas
        energia -= energia * (self.perdidas / 100)

        # Eficiencia
        energia *= self.eficiencia / 100

        # Temperatura
        if self.temperatura > 25:

            grados_extra = self.temperatura - 25

            perdida = grados_extra * 0.4

            energia *= (1 - perdida / 100)

        return energia
```

### Correcto 

```python
from abc import ABC, abstractmethod


class GeneracionSolar(ABC):

    @abstractmethod
    def calcular(self):
        pass


class GeneracionBase(GeneracionSolar):
    """
    Generación básica:
    Energía = Potencia × Horas de sol
    """

    def __init__(self, potencia, horas_sol):
        self.potencia = potencia
        self.horas_sol = horas_sol

    def calcular(self):
        return self.potencia * self.horas_sol


class GeneracionDecorator(GeneracionSolar):

    def __init__(self, generacion):
        self.generacion = generacion

    def calcular(self):
        return self.generacion.calcular()


class PerdidasDecorator(GeneracionDecorator):

    def __init__(self, generacion, porcentaje):
        super().__init__(generacion)
        self.porcentaje = porcentaje

    def calcular(self):
        energia = self.generacion.calcular()

        perdida = energia * (self.porcentaje / 100)

        return energia - perdida


class EficienciaDecorator(GeneracionDecorator):

    def __init__(self, generacion, eficiencia):
        super().__init__(generacion)
        self.eficiencia = eficiencia

    def calcular(self):
        energia = self.generacion.calcular()

        return energia * (self.eficiencia / 100)


class TemperaturaDecorator(GeneracionDecorator):

    def __init__(self, generacion, temperatura):
        super().__init__(generacion)
        self.temperatura = temperatura

    def calcular(self):
        energia = self.generacion.calcular()

        if self.temperatura > 25:

            grados_extra = self.temperatura - 25

            perdida = grados_extra * 0.4

            energia *= (1 - perdida / 100)

        return energia
```

## Bridge
"El patrón Bridge separa la abstracción SistemaMonitoreado de su implementación Monitor. De esta manera puedo cambiar entre consola, web o API sin modificar la clase principal."

**Justificación vs Alternativas:** ¿Por qué no usar condicionales simples (`if tipo == 'web': ...`) o herencia múltiple? Los condicionales rompen el principio de responsabilidad única al llenar una sola clase con todos los medios de envío. Al separar la abstracción (tipo de sistema: residencial/comercial) de la implementación (monitor: web/api/consola), ambas jerarquías pueden crecer (ej. agregar un monitor móvil) sin multiplicarse en `SistemaResidencialWeb`, `SistemaComercialWeb`, etc.

### incorrecto 

```python
class SistemaMonitoreado:

    def mostrar_estado(self, mensaje, tipo):

        if tipo == "consola":
            print(f"[CONSOLA] {mensaje}")

        elif tipo == "web":
            print(f"[WEB] {mensaje}")

        elif tipo == "api":
            print(f"[API] {mensaje}")


sistema = SistemaMonitoreado()

sistema.mostrar_estado("Sistema funcionando", "consola")
sistema.mostrar_estado("Sistema funcionando", "web")
```

### Correcto

```python
from abc import ABC, abstractmethod

class Monitor(ABC):
    @abstractmethod
    def mostrar(self, mensaje):
        pass

class MonitorConsola(Monitor):
    def mostrar(self, mensaje):
        print(f"[CONSOLA] {mensaje}")

class MonitorWeb(Monitor):
    def mostrar(self, mensaje):
        print(f"[WEB] {mensaje}")

class MonitorAPI(Monitor):
    def mostrar(self, mensaje):
        print(f"[API] {mensaje}")

class SistemaMonitoreado(ABC):
    def __init__(self, monitor):
        self.monitor = monitor
    def cambiar_monitor(self, monitor):
        self.monitor = monitor
    @abstractmethod
    def mostrar_estado(self, mensaje):
        pass

class SistemaResidencial(SistemaMonitoreado):
    def mostrar_estado(self, mensaje):
        self.monitor.mostrar(f"🏠 [Residencial] {mensaje}")

class SistemaComercial(SistemaMonitoreado):
    def mostrar_estado(self, mensaje):
        self.monitor.mostrar(f"🏢 [Comercial] URGENTE: {mensaje}")

# Ejemplo de uso
monitor = MonitorConsola()
sistema = SistemaResidencial(monitor)
sistema.mostrar_estado("Sistema funcionando")
sistema.cambiar_monitor(MonitorWeb())
sistema.mostrar_estado("Sistema funcionando")
```
