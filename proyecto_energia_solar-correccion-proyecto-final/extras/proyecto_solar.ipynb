### Problema
En el sistema de energía solar, diferentes módulos necesitan acceder a la configuración de la API. Si cada módulo crea su propia configuración, pueden existir datos duplicados o inconsistentes.

```python
class APIConfig:
    
    def __init__(self):
        self.api_url = "https://api.open-meteo.com/v1/forecast"
        self.unidad_energia = "Wh"
        self.unidad_temperatura = "°C"

    def mostrar_configuracion(self):
        return {
            "API": self.api_url,
            "Energía": self.unidad_energia,
            "Temperatura": self.unidad_temperatura
        }


# Se crean dos objetos diferentes
config1 = APIConfig()
config2 = APIConfig()

print(config1 is config2)
```

### Solucion 
Se implementó el patrón Singleton para garantizar que exista una única instancia de APIConfig. De esta manera, todos los módulos del sistema comparten la misma URL de la API y las mismas unidades de energía y temperatura.

La configuración queda centralizada, evitando duplicación y manteniendo consistencia en todo el sistema.

```python
class APIConfig:
    """
    Patrón Singleton.
    Solo puede existir una instancia de esta clase.
    """

    _instancia = None

    def __new__(cls):
        if cls._instancia is None:
            cls._instancia = super().__new__(cls)

            cls._instancia.api_url = (
                "https://api.open-meteo.com/v1/forecast"
            )

            cls._instancia.unidad_energia = "Wh"
            cls._instancia.unidad_temperatura = "°C"

        return cls._instancia

    def mostrar_configuracion(self):
        return {
            "API": self.api_url,
            "Energía": self.unidad_energia,
            "Temperatura": self.unidad_temperatura
        }
```

### Problema 
En un sistema de monitoreo o diseño de energía solar pueden existir diferentes instalaciones con configuraciones muy similares. Crear cada sistema desde cero implica repetir información y código.

```python
class SistemaSolar:

    def __init__(
        self,
        nombre,
        cantidad_paneles,
        potencia_panel,
        tipo_panel,
        ubicacion
    ):
        self.nombre = nombre
        self.cantidad_paneles = cantidad_paneles
        self.potencia_panel = potencia_panel
        self.tipo_panel = tipo_panel
        self.ubicacion = ubicacion


sistema_original = SistemaSolar(
    "Sistema Solar Ciénaga",
    10,
    550,
    "Monocristalino",
    "Ciénaga, Magdalena"
)

sistema_copia = SistemaSolar(
    "Sistema Solar Santa Marta",
    10,
    550,
    "Monocristalino",
    "Santa Marta, Magdalena"
)
```

### Solucion 
Se utiliza el patrón de diseño Prototype, que permite crear nuevos sistemas solares clonando un objeto existente. De esta manera, se reutiliza la configuración inicial y solamente se modifican las características que sean diferentes

La clase SistemaSolar representa el prototipo y el método clonar() utiliza copy.deepcopy() para generar una copia independiente del sistema original.

Se reduce la repetición de código y se facilita la creación de múltiples configuraciones de sistemas solares similares.

```python
import copy


class SistemaSolar:
    """
    Patrón Prototype.
    Permite clonar un sistema solar existente.
    """

    def __init__(
        self,
        nombre,
        cantidad_paneles,
        potencia_panel,
        tipo_panel,
        ubicacion
    ):
        self.nombre = nombre
        self.cantidad_paneles = cantidad_paneles
        self.potencia_panel = potencia_panel
        self.tipo_panel = tipo_panel
        self.ubicacion = ubicacion

    def clonar(self):
        return copy.deepcopy(self)

    def calcular_potencia(self):
        return (
            self.cantidad_paneles
            * self.potencia_panel
        )

    def mostrar_informacion(self):
        return (
            f"Nombre: {self.nombre}\n"
            f"Paneles: {self.cantidad_paneles}\n"
            f"Potencia por panel: {self.potencia_panel} W\n"
            f"Tipo: {self.tipo_panel}\n"
            f"Ubicación: {self.ubicacion}\n"
            f"Potencia total: "
            f"{self.calcular_potencia()} W"
        )
```

### problema 
Era necesario manejar paneles solares individuales y grupos de paneles. Si se trataban de forma diferente, el código podía volverse más complejo y difícil de mantener.

```python
class PanelSolar:

    def __init__(self, identificador, potencia, tipo):
        self.identificador = identificador
        self.potencia = potencia
        self.tipo = tipo


class ArregloSolar:

    def __init__(self, nombre):
        self.nombre = nombre
        self.paneles = []

    def agregar(self, panel):
        self.paneles.append(panel)

    def calcular_potencia(self):
        total = 0

        for panel in self.paneles:
            total += panel.potencia

        return total

    def mostrar(self):
        print(f"Arreglo: {self.nombre}")

        for panel in self.paneles:
            print(
                f"Panel {panel.identificador}: "
                f"{panel.potencia} W - {panel.tipo}"
            )

        print(f"Potencia total: {self.calcular_potencia()} W")
```

### Solucion 
Se utiliza el patrón Composite, creando una clase común ComponenteSolar. Así, tanto PanelSolar como ArregloSolar pueden utilizar los mismos métodos: calcular_potencia() y mostrar().

El sistema puede calcular y mostrar la información de un panel o de un arreglo completo de manera uniforme y sencilla.

```python
from abc import ABC, abstractmethod


class ComponenteSolar(ABC):

    @abstractmethod
    def calcular_potencia(self):
        pass

    @abstractmethod
    def mostrar(self):
        pass


class PanelSolar(ComponenteSolar):
    """
    Hoja del patrón Composite.
    """

    def __init__(self, identificador, potencia, tipo):
        self.identificador = identificador
        self.potencia = potencia
        self.tipo = tipo

    def calcular_potencia(self):
        return self.potencia

    def mostrar(self):
        return (
            f"Panel {self.identificador}: "
            f"{self.potencia} W - {self.tipo}"
        )


class ArregloSolar(ComponenteSolar):
    """
    Composite.
    Puede contener varios paneles.
    """

    def __init__(self, nombre):
        self.nombre = nombre
        self.componentes = []

    def agregar(self, componente):
        self.componentes.append(componente)

    def eliminar(self, componente):
        if componente in self.componentes:
            self.componentes.remove(componente)

    def calcular_potencia(self):
        total = 0

        for componente in self.componentes:
            total += componente.calcular_potencia()

        return total

    def mostrar(self):
        resultado = []

        resultado.append(f"\nArreglo: {self.nombre}")

        for componente in self.componentes:
            resultado.append(componente.mostrar())

        resultado.append(
            f"Potencia total: {self.calcular_potencia()} W"
        )

        return "\n".join(resultado)
```

### Problema 
"El problema que encontramos es que la generación de energía solar puede verse afectada por diferentes factores, como pérdidas, eficiencia y temperatura. Si colocáramos toda esta lógica dentro de una sola clase, el código se volvería difícil de mantener y tendríamos que modificar la clase cada vez que apareciera un nuevo factor.

```python
class GeneracionBase:

    def __init__(
        self,
        potencia,
        horas_sol,
        perdidas=0,
        eficiencia=100,
        temperatura=25
    ):
        self.potencia = potencia
        self.horas_sol = horas_sol
        self.perdidas = perdidas
        self.eficiencia = eficiencia
        self.temperatura = temperatura

    def calcular(self):

        energia = self.potencia * self.horas_sol

        # Pérdidas
        energia -= energia * (self.perdidas / 100)

        # Eficiencia
        energia *= self.eficiencia / 100

        # Temperatura
        if self.temperatura > 25:

            grados_extra = self.temperatura - 25

            perdida = grados_extra * 0.4

            energia *= (1 - perdida / 100)

        return energia
```

### solución
La solución fue utilizar el patrón Decorator, que nos permite agregar estas condiciones de manera dinámica sin modificar la clase principal de generación. Así podemos combinar diferentes decoradores según las condiciones que queramos evaluar."

```python
from abc import ABC, abstractmethod


class GeneracionSolar(ABC):

    @abstractmethod
    def calcular(self):
        pass


class GeneracionBase(GeneracionSolar):
    """
    Generación básica:
    Energía = Potencia × Horas de sol
    """

    def __init__(self, potencia, horas_sol):
        self.potencia = potencia
        self.horas_sol = horas_sol

    def calcular(self):
        return self.potencia * self.horas_sol


class GeneracionDecorator(GeneracionSolar):

    def __init__(self, generacion):
        self.generacion = generacion

    def calcular(self):
        return self.generacion.calcular()


class PerdidasDecorator(GeneracionDecorator):

    def __init__(self, generacion, porcentaje):
        super().__init__(generacion)
        self.porcentaje = porcentaje

    def calcular(self):
        energia = self.generacion.calcular()

        perdida = energia * (self.porcentaje / 100)

        return energia - perdida


class EficienciaDecorator(GeneracionDecorator):

    def __init__(self, generacion, eficiencia):
        super().__init__(generacion)
        self.eficiencia = eficiencia

    def calcular(self):
        energia = self.generacion.calcular()

        return energia * (self.eficiencia / 100)


class TemperaturaDecorator(GeneracionDecorator):

    def __init__(self, generacion, temperatura):
        super().__init__(generacion)
        self.temperatura = temperatura

    def calcular(self):
        energia = self.generacion.calcular()

        if self.temperatura > 25:

            grados_extra = self.temperatura - 25

            perdida = grados_extra * 0.4

            energia *= (1 - perdida / 100)

        return energia
```

### Problema
El sistema de monitoreo necesitaba mostrar el estado de la instalación solar en diferentes medios, como consola, web o API.

Si el sistema estuviera directamente programado para cada tipo de monitor, sería necesario modificar el código principal cada vez que se quisiera agregar o cambiar un medio de visualización. Esto generaría dependencia entre el sistema y el tipo de monitor, haciendo el código más difícil de mantener y ampliar.

```python
class SistemaMonitoreado:

    def mostrar_estado(self, mensaje, tipo):

        if tipo == "consola":
            print(f"[CONSOLA] {mensaje}")

        elif tipo == "web":
            print(f"[WEB] {mensaje}")

        elif tipo == "api":
            print(f"[API] {mensaje}")


sistema = SistemaMonitoreado()

sistema.mostrar_estado("Sistema funcionando", "consola")
sistema.mostrar_estado("Sistema funcionando", "web")
```

### Solucion 
Aplicamos Bridge para separar el sistema (SistemaResidencial, SistemaComercial) del monitor (MonitorConsola, MonitorWeb, MonitorAPI).

Podemos cambiar el tipo de monitor sin modificar la lógica del sistema, haciendo el código más flexible, reutilizable y fácil de mantener.

```python
from abc import ABC, abstractmethod

class Monitor(ABC):
    @abstractmethod
    def mostrar(self, mensaje):
        pass

class MonitorConsola(Monitor):
    def mostrar(self, mensaje):
        print(f"[CONSOLA] {mensaje}")

class MonitorWeb(Monitor):
    def mostrar(self, mensaje):
        print(f"[WEB] {mensaje}")

class MonitorAPI(Monitor):
    def mostrar(self, mensaje):
        print(f"[API] {mensaje}")

class SistemaMonitoreado(ABC):
    def __init__(self, monitor):
        self.monitor = monitor
    def cambiar_monitor(self, monitor):
        self.monitor = monitor
    @abstractmethod
    def mostrar_estado(self, mensaje):
        pass

class SistemaResidencial(SistemaMonitoreado):
    def mostrar_estado(self, mensaje):
        self.monitor.mostrar(f"🏠 [Residencial] {mensaje}")

class SistemaComercial(SistemaMonitoreado):
    def mostrar_estado(self, mensaje):
        self.monitor.mostrar(f"🏢 [Comercial] URGENTE: {mensaje}")

# Ejemplo de uso
monitor = MonitorConsola()
sistema = SistemaResidencial(monitor)
sistema.mostrar_estado("Sistema funcionando")
sistema.cambiar_monitor(MonitorWeb())
sistema.mostrar_estado("Sistema funcionando")
```
